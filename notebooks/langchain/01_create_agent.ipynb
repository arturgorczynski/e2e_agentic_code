# First LangChain agent

An agent receives messages, calls a language model, and returns an assistant response. Without a checkpointer, separate invocations do not retain earlier conversation turns.

[LangChain agents](https://docs.langchain.com/oss/python/langchain/agents)

```python
# Imports and project path setup only; examples start in the next cell.
from pathlib import Path
import sys

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "models.toml").exists())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from langchain.agents import create_agent
from model_config import get_json_model, get_model, get_model_name
```

```python
model = get_model()
agent = create_agent(
    model=model,
    tools=[],
    system_prompt="You are a helpful, concise assistant Joe the Bartender",
)
```

```python
result = agent.invoke({
    "messages": [{"role": "user", "content": "Explain what an agent is in one sentence."}]
})
print(result["messages"][-1].content)
```

```python
result = agent.invoke({
    "messages": [{"role": "user", "content": "What was my last question?"}]
})
print(result["messages"][-1].content)
```

```python
result = agent.invoke({
    "messages": [{"role": "user", "content": "Who are you?"}]
})
print(result["messages"][-1].content)
```
