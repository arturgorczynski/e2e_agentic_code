# A conversational input loop

A Python loop accepts successive questions and invokes the agent for each one. Input validation rejects empty requests; an explicit stop command ends the loop. Reusing the agent object alone does not preserve conversation history.

[LangChain agents](https://docs.langchain.com/oss/python/langchain/agents)

```python
# Imports and project path setup only; examples start in the next cell.
from pathlib import Path
import sys

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "models.toml").exists())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from langchain.agents import create_agent
from model_config import get_json_model, get_model, get_model_name
```

```python
model = get_model()


agent = create_agent(
    model=model,
    tools=[],
    system_prompt="You are a helpful, concise assistant.",
)
```

```python
def ask_question(prompt:str) -> str:

    if len(prompt.strip()) > 3:
        result = agent.invoke({
            "messages": [{"role": "user", "content": prompt}]
        })
        answer = result["messages"][-1].content

        print(answer)
        return answer

    else:
        return "Invalid input prompt "
```

```python
while True:

    request = input("Ask Agent: ")

    if request in ["exit", "q", "quit", "stop"]:
        break
    else:
        ask_question(request)
```
